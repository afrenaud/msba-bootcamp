### Distribution - Box plot

We want to check if there is any difference in the `daily_unit_sales` and `daily_unit_sales_kg` for Product 6. Create a boxplot for both of these columns for all three years side-by-side and compare the results.

Hint: You can create a new smaller dataframe by filtering out columns for just Product 6.

```python
df['Year'] = df['Date'].dt.year
p6 = df[df['product_id']==6]

fig, axes = plt.subplots(1, 2, figsize=(12,6))

sns.boxplot(data=p6, x='Year', y='daily_unit_sales', ax=axes[0])
sns.boxplot(data=p6, x='Year', y='daily_unit_sales_kg', ax=axes[1])

plt.tight_layout()
```

```python
# Your code goes here
#disp(hint + 'Mevgal-06')
```

Before we go into the analysis of these plots, let's make sure the axes are labelled right and that our plots have a proper title.

Add the following to the above chart:

*   Set the titles for both the graphs as `Daily Sales - Product 6` and `Daily Sales (kg) - Product 6` respectively
*   Set the y-axis labels as `Daily Sales (units)` and `Daily Sales (kg)` respectively
*   Add a title for both charts - `Daily Sales Comparison - Product 6`
*   Look through the different `color_palette` options offered by Seaborn and add a little colour to your graph to differentiate various years

Hint: `suptitle` function is used to define a common title for all subplots.

```python
p6 = df[df['product_id']==6]

fig, axes = plt.subplots(1, 2, figsize=(12,6))

sns.boxplot(data=p6, x='Year', y='daily_unit_sales', ax=axes[0], hue='Year', palette=sns.color_palette("Set2", p6['Year'].nunique()))
axes[0].set_ylabel('Daily Sales (units)')
axes[0].set_xticks([])  # Remove x-ticks
axes[0].set_xlabel('')  # Remove x-label

sns.boxplot(data=p6, x='Year', y='daily_unit_sales_kg', ax=axes[1], hue='Year', palette=sns.color_palette("Set2", p6['Year'].nunique()))
axes[1].set_ylabel('Daily Sales (kg)')
axes[1].set_xticks([])  # Remove x-ticks
axes[1].set_xlabel('')  # Remove x-label

plt.suptitle('Daily Sales Comparison - Product 6 (2020-2022)', y=1)
plt.tight_layout();
```

This is way too busy and we cannot understand any patterns. Let's aggregate to the month level and then try again.

First, create a Month column than includes both month and year:

Hint: Use `.dt.to_period('M')` to create monthly frequency. Then use `.dt.to_timestamp()` to convert this monthly object into a date object, otherwise, you will not be able to utilize some of the seaborn/matplotlib automatic date handling.

```python
df['Month'] = df['Date'].dt.to_period('M').dt.to_timestamp()
df[['Month']].head(2)
```

```python
# Your code goes here
#disp(hint + 'Mevgal-11')
```

Then, for each product, aggregate the data to month level (year-month-product level), using `groupby()`. Use `sum()` as the aggregate function. Call the resulting DataFrame `monthly_aggregated_df`.

Discussion: what are the downsides and upsides of `sum()` in this scenario?

```python
monthly_aggregated_df = df.groupby(['Month', 'product_id'])['daily_unit_sales_kg'].sum().reset_index()
monthly_aggregated_df.head(3)
```

```python
# Your code goes here
#disp(hint + 'Mevgal-12')
```

Now plot the monthly `daily_unit_sales_kg` from `monthly_aggregated_df` as a lineplot, with one line per product. Add a title and axis labels.

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=monthly_aggregated_df, x='Month', y='daily_unit_sales_kg', hue='product_id', marker='o', palette="Paired")

plt.title('Monthly Aggregated Daily Unit Sales (kg)')
plt.xlabel('Date')
plt.ylabel('Daily Unit Sales (kg)')
plt.xticks(rotation=90)
plt.tight_layout()
```
